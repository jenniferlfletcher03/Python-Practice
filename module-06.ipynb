# Lesson 2: `and`, `or`, `not`

These are the logical operators that combine booleans:
- `A and B` → True only if BOTH are True
- `A or B` → True if EITHER is True (or both)
- `not A` → flips True to False and vice versa

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
is_morning = True
has_coffee = False

print(is_morning and has_coffee)   # both must be true
print(is_morning or has_coffee)    # either is enough
print(not has_coffee)              # flip it
print(is_morning and not has_coffee)  # combine
````

# Lesson 3: Truthy and Falsy Values

Python doesn't just treat `True` and `False` as booleans in conditional contexts. It treats *any value* as either **truthy** or **falsy**.

**Falsy values** (treated as False in `if` statements):
- `False`
- `None`
- `0` (the integer zero)
- `""` (the empty string)
- `[]` (the empty list)
- `{}` (the empty dict)

**Everything else is truthy** — including non-zero numbers, non-empty strings, non-empty lists, etc.

### 3a — Falsy values in `if` statements

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
def check(value):
    if value:
        print(f"{value!r} is truthy")
    else:
        print(f"{value!r} is falsy")

check(True)
check(False)
check(None)
check(0)
check("")
check([])
check("hello")
check([1, 2])
check(42)
````

# Lesson 4: Short-Circuit Evaluation — the `x or default` idiom

This is where this morning's `joined or None` line FINALLY clicks.

When Python evaluates `A or B`:
- If A is truthy, the whole expression is A (B is never even evaluated)
- If A is falsy, the whole expression is B

So `joined or None` reads in plain English: *"give me `joined` if it's truthy; otherwise give me `None`."*

### 4a — See it in action

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
joined = "Hello world"
result = joined or None
print(f"When joined is non-empty: {result!r}")

joined = ""
result = joined or None
print(f"When joined is empty: {result!r}")
````

### 4b — The pattern generalizes

`some_value or fallback` is how you say *"use some_value if it's meaningful, otherwise use the fallback."*

````python
user_input = ""
name = user_input or "Anonymous"
print(f"Hello, {name}!")

user_input = "Jen"
name = user_input or "Anonymous"
print(f"Hello, {name}!")
````
