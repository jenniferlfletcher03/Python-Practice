# Lesson 2: `isinstance()` — Checking Types Properly

`type()` *tells* you what type something is. `isinstance()` *asks* whether something is a given type and returns True or False. `isinstance()` is what you actually use in real code, because you usually want to *decide* based on type, not just inspect it.

This is the same `isinstance()` we used in `extract_text` to check whether `content` was a string or a list.

# Lesson 3: `None` — Python's *"Nothing Here"* Value

`None` is its own type. It's not zero, not an empty string, not an empty list. It's the value Python uses to mean *"this slot is intentionally empty."*

Functions that don't explicitly `return` something return `None` automatically. Missing dict lookups with `.get()` return `None`. Many things return `None`.

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
def says_nothing():
    pass    # function does nothing, no return statement

result = says_nothing()
print(result)
print(type(result))
print(result is None)
print(result == None)
````

### 3b — Always check for `None` with `is`, not `==`

Both `result is None` and `result == None` happen to return True in the example above. But the Python convention is to ALWAYS use `is None` (and `is not None`) when checking for None. Reason: `is` checks for *identity* (literally the same object). There's only ever ONE `None` in a running Python program, so identity-checking is the right semantic. Some object types define funny `__eq__` methods that can make `== None` behave unexpectedly.

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
value = None

if value is None:
    print("value is None — handle the empty case")

value = "Jen"

if value is not None:
    print(f"value is something: {value}")
````

### Now compare to a reference implementation

Below is one way to write `to_text`. **Predict** what it does, run it, then compare to YOUR version.

````python
def to_text(content):
    """Normalize content (str / list / None / other) into a single text string."""
    if content is None:
        return ""
    if isinstance(content, str):
        return content.strip()
    if isinstance(content, list):
        return "\n".join(content)
    return str(content)

print(repr(to_text(None)))
print(repr(to_text("  Hello world  ")))
print(repr(to_text(["line one", "line two"])))
print(repr(to_text(42)))
print(repr(to_text({"key": "v"})))
````

Look at the order of the checks: `is None` first, then `isinstance(content, str)`, then `isinstance(content, list)`, then a final `return str(content)` as a catch-all. Why does order matter here?
