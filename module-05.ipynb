# Lesson 1: Defining Your Own Function

The `def` keyword tells Python *"I'm defining a function."* The pattern is:

```python
def function_name(arguments):
    # indented body — the code that runs when the function is called
    return some_value
```

Defining a function doesn't *run* the code. It just teaches Python what to do when you later call the function by name.

# Lesson 2: Return Values

`print()` *displays* something. `return` *hands a value back to whoever called the function*. These are completely different operations — and confusing them is one of the most common beginner mistakes. (We worked through this exact distinction with `.get()` in Module 04.)

### 2a — Print vs return

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
def add_print(a, b):
    print(a + b)  # displays it

def add_return(a, b):
    return a + b  # hands it back

# Calling add_print: you see output, but the function gives back nothing usable
result1 = add_print(2, 3)
print(f"result1 is: {result1}")

# Calling add_return: no automatic output, but the function gives back a value to store
result2 = add_return(2, 3)
print(f"result2 is: {result2}")
````

**📝 Reflect** — what did I actually see, and what did I learn?

*(write your reflection here, in your own words, after running)*

### 2b — A function that returns is reusable in expressions

Because `add_return` hands back a value, you can use the call *anywhere a value would normally go*.

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
total = add_return(5, 7) + add_return(10, 20)
print(f"Total: {total}")
````

# Lesson 3: Default Arguments

Sometimes you want an argument to have a sensible default value when the caller doesn't supply one. Python lets you specify defaults in the function definition. This is exactly how `.get(key, default)` works.

### 3a — A function with a default argument

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
def greet_with_mood(name, mood="cheerful"):
    return f"Hello {name}, hope you're feeling {mood} today!"

print(greet_with_mood("Jen"))                    # uses the default
print(greet_with_mood("Jen", mood="thoughtful"))  # overrides the default
````

# Lesson 4: Capstone — Write a Useful Function

Now we put the pieces together. Write a function that takes a dictionary and a key, and safely returns the value — printing a friendly message if the key is missing.

*(This is essentially the inner logic of `.get()`, written by you.)*

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
def safe_lookup(some_dict, key, fallback="not set"):
    """Return the value for key if it exists, otherwise the fallback."""
    if key in some_dict:
        return some_dict[key]
    else:
        return fallback

# Test it on your learning_today dict
learning_today = {
    "module": 5,
    "topic": "functions",
    "feelings": "more confident"
}

print(safe_lookup(learning_today, "module"))
print(safe_lookup(learning_today, "favorite_color"))
print(safe_lookup(learning_today, "favorite_color", fallback="I haven't decided"))
````
