# Lesson 1: Defining Your Own Function

The `def` keyword tells Python *"I'm defining a function."* The pattern is:

```python
def function_name(arguments):
    # indented body — the code that runs when the function is called

```

Defining a function doesn't *run* the code. It just teaches Python what to do when you later call the function by name.

# Lesson 2: Return Values

`print()` *displays* something. `return` *hands a value back to whoever called the function*. These are completely different operations — and confusing them is one of the most common beginner mistakes. (We worked through this exact distinction with `.get()` in Module 04.)

### 2a — Print vs return

**🔮 Predict** — what do I think the next code cell will do?

Okay...the function add_print(a, b) is going to display "a + b" so for result1 I think it's going to look like this:

result1 is: 2 + 3

the add_return function I *think* is actually going to add the numbers and hold the result, so that when you call a print on result2 it should print like this:

result2 is: 5

````python
def add_print(a, b):
    print(a + b)  # displays it

def add_return(a, b):
    return a + b  # hands it back

# Calling add_print: you see output, but the function gives back nothing usable
result1 = add_print(2, 3)
print(f"result1 is: {result1}")

# Calling add_return: no automatic output, but the function gives back a value to store
result2 = add_return(2, 3)
print(f"result2 is: {result2}")
````

I had to talk with Claude to nail this down, but the reason result1 returned "None" was because the print command didn't grab the math from 2 + 3...python did it and dropped it because print is just a display function and not a storing function like return. the add_return(a, b) function actually gave the result back to result2 where it could store it as the variable until the print command was called for it.

````python
x = print("Hi")
print(f"x is: {x}")
````

# Lesson 3: Default Arguments

Sometimes you want an argument to have a sensible default value when the caller doesn't supply one. Python lets you specify defaults in the function definition. This is exactly how `.get(key, default)` works.

### 3a — A function with a default argument

**🔮 Predict** — what do I think the next code cell will do?

Okay...this function greet_with_mood(name, mood="cheerful") sets the default mood to cheeful, so that if one isn't provided that what the function pulls. That means that print(greet_with_mood("Jen")) will print:

Hello Jen, hope you're feeling cheerful today!"

Where as, when the mood is specified, the function will override the default and print:

Hello Jen, hope you're feeling thoughtful today!

````python
def greet_with_mood(name, mood="cheerful"):
    return f"Hello {name}, hope you're feeling {mood} today!"

print(greet_with_mood("Jen"))                    # uses the default
print(greet_with_mood("Jen", mood="thoughtful"))  # overrides the default
````

**📝 Reflect** — what did I actually see, and what did I learn?

Defaults are handy...they give a pre-written response incase there isn't variable provided.

**Spec — your function `safe_lookup`:**
- Takes three arguments: `some_dict`, `key`, and `fallback` (with a default of `"not set"`)
- If the key exists in `some_dict`, **return** the value at that key
- If the key does NOT exist, **return** the fallback

You can use `in` to check whether a key is in a dict — that's `if key in some_dict:`.

````python
# your safe_lookup here, then the test calls
learning_today = {
    "module": 5,
    "topic": "functions",
    "feelings": "more confident"
}

def safe_lookup(dictionary, key, fallback = "not set"):
    if key in learning_today:
        return learning_today[key]
    else:
        return fallback

print(safe_lookup(learning_today, "module"))
print(safe_lookup(learning_today, "favorite_color"))
print(safe_lookup(learning_today, "favorit_color", fallback = "I haven't decided"))


````
