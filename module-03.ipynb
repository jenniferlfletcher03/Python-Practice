# Lesson 1

Lesson 1 of module 3 introduced how to have python read and store data that you give it out of markdown or text files using this code block:

f = open("email.txt", "r")
email = f.read()
f.close()  

This block has python open the file email.txt as a read file ("r") and stores it in the variable f. f.read() actually reads the file and then f.close() closes the file so that it doesn't continue to take up memory and crowd out other files. If you were to write on this file ("w") you also run the potential of loosing your progress as well if you don't close.
