### Reference implementation

**Predict** what this prints before you run.

```python
def summarize_text_file(path):
    with open(path) as f:
        content = f.read()

    char_count = len(content)
    word_count = len(content.split())
    first_line = content.splitlines()[0] if content else "(empty file)"

    print(f"path: {path}")
    print(f"characters: {char_count}")
    print(f"words: {word_count}")
    print(f"first line: {first_line}")

summarize_text_file("/Users/jenniferfletcher/Desktop/Python-Practice/README.md")
```
