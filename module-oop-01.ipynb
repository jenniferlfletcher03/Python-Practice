# Module OOP-01: What a "class" even *is*

Up to now, almost everything you've written has used Python's *built-in* shapes — strings, lists, dicts, functions. Those are great, but there's a shape you haven't used yet that the rest of the OOP curriculum is built on: a **class**.

A class is a **blueprint**.
An **instance** is a thing made *from* that blueprint.

### The sewing analogy 🧵

A sewing pattern is a class. The pattern says *"this garment has a bodice, two sleeves, a collar of this shape."* It is structure — not fabric, not a finished garment.

Each garment you cut and sew from that pattern is an **instance**. They all share the structure the pattern defines, but each one is its own thing: different fabric, different size, different person it was made for.

In code, you `class Account:` to define the structure. Then every time you write `Account(...)` you cut a new garment from that pattern. That new garment is an instance.

> **What this module is for:** the *Account* class is just a substrate. The real skill you're building is the *move* — defining a blueprint and creating instances from it. You'll see at the end that the same move works for `SewingPattern`, `ChoirMember`, or `RoutingNode` just as well. Bank-account is what the Pre-Screen assessment uses — that's why we start there — but the instinct generalizes.

# Lesson 1: A class is a blueprint

The simplest possible class. Read it, then **predict** what each `print` line will show — *especially* the last one (the `is` comparison).


````python
class Account:
    bank_name = "Bank of Jen"

# Cut two garments from the same pattern.
acct1 = Account()
acct2 = Account()

# They share the structure the pattern defines...
print("acct1's bank:", acct1.bank_name)
print("acct2's bank:", acct2.bank_name)

# ...but they are not the same garment.
print("are they the same object?", acct1 is acct2)
````

**Reflect:**

- Both accounts said `"Bank of Jen"` — that's the *pattern* speaking. Every Account, by definition, lives at Bank of Jen.
- But `acct1 is acct2` was `False`. They are two different objects in memory. Same pattern, different garments.

The thing that surprises most beginners: `Account` (capital A, the class) and `acct1` (a specific instance) are **different things**. The class is the pattern. The instance is the cut-and-sewn garment.


# Lesson 2: `__init__` — the moment of creation

A class with no `__init__` produces identical instances — every Account looks the same. That's not useful. We want each account to have its *own* owner and *own* balance.

`__init__` is the **constructor** — a special method Python runs the moment you create a new instance. Sewing analogy: it's the step where you take measurements and pick fabric before cutting. The pattern is shared; the customization happens at creation.

`self` is the new instance's way of referring to *itself* during setup. It feels weird at first; it clicks fast.

**Predict** what each `print` will show — and *why* the two accounts don't interfere with each other.


````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

# Two accounts, customized at the moment of creation.
jen_acct = Account("Jen", 100)
sam_acct = Account("Sam", 500)

print(jen_acct.owner, "has", jen_acct.balance)
print(sam_acct.owner, "has", sam_acct.balance)
````

**Reflect:**

`self.owner = owner` means *"on this specific instance, set its owner attribute to whatever was passed in."* Every instance keeps its own data — like two garments cut from the same pattern but in different fabrics. Same shape, different identity.

If `self` still feels weird, here's the move: every time you see `self.x` inside a class, mentally read it as *"this instance's x."* It's just the instance's name for itself, while it's being built.


# Lesson 3: Methods — what an instance can *do*

Attributes (`owner`, `balance`) are what an instance *has*. **Methods** are what an instance *can do*: deposit money, withdraw money.

A method is just a function defined inside the class. First argument is always `self` — same trick as `__init__`. When you call `jen_acct.deposit(50)`, Python automatically passes `jen_acct` in as `self`.

Sewing analogy: methods are the *operations* a garment supports — *"hem the sleeve," "let out the waist."* Things you do to a specific garment, not to the pattern itself.

**Predict** what the four `print` lines will show.


````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount

jen_acct = Account("Jen", 100)
print("starting balance:", jen_acct.balance)

jen_acct.deposit(50)
print("after deposit:", jen_acct.balance)

jen_acct.withdraw(30)
print("after withdraw:", jen_acct.balance)

# A second account — should be untouched by what we did to Jen's.
sam_acct = Account("Sam", 500)
print("Sam's balance:", sam_acct.balance)
````

**Reflect:**

Two important things just happened:

1. **You never passed `self` yourself.** When you wrote `jen_acct.deposit(50)`, Python silently rewrote it as `Account.deposit(jen_acct, 50)`. That's why `self` inside the method points at the specific account you called it on.
2. **`sam_acct` was unaffected by Jen's deposits and withdrawals.** Each instance has its own `balance` attribute. Two garments, two separate bodies of fabric.
