# Lesson 1: Escape Sequences

An **escape sequence** is a special two-character code that represents a single character that's hard to type directly. The backslash `\` is the *escape character* — it tells Python *"the next character has a special meaning."*

Most common escape sequences:
- `\n` — newline (line break)
- `\t` — tab
- `\\` — a literal backslash
- `\"` — a literal double-quote (inside a double-quoted string)
- `\'` — a literal single-quote (inside a single-quoted string)

# Lesson 2: `.strip()`, `.upper()`, `.lower()` — Cleaning Methods

These are methods on strings that return a *new* string with some transformation applied. Important: strings are **immutable** — methods don't change the original, they hand back a new version.

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

````python
messy = "   Hello World   \n"

print(f"Original: {messy!r}")
print(f"Stripped: {messy.strip()!r}")
print(f"Upper:    {messy.upper()!r}")
print(f"Lower:    {messy.lower()!r}")

# The original is unchanged — methods return NEW strings
print(f"After all that, messy is still: {messy!r}")
````

### Now compare to a reference implementation

Below is one way to write `clean_tags`. **Predict** what it does (it should match yours), run it, then compare to YOUR version. Where they differ — is one better, or are they just different styles?

````python
def clean_tags(tag_string):
    """Split a comma-separated tag string, strip + lowercase each piece, drop the empties."""
    raw_pieces = tag_string.split(",")
    cleaned = []
    for piece in raw_pieces:
        cleaned_piece = piece.strip().lower()
        if cleaned_piece:        # ← the truthy/falsy idiom from Module 06
            cleaned.append(cleaned_piece)
    return cleaned

print(clean_tags("  Python , AI ,memory,  NOTES  "))
print(clean_tags("thesis,alignment,bidirectional"))
print(clean_tags("  ,   ,a,b,  "))
````

**📝 Reflect** — how does the reference compare to yours? Notice the `if cleaned_piece:` line — that's the **truthy/falsy** idiom from Module 06 doing real work: skipping any piece that strips down to an empty string. Did your version drop the empties? If not, you can see now how a single `if` line handles it.
