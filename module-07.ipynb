# Lesson 1: Escape Sequences

An **escape sequence** is a special two-character code that represents a single character that's hard to type directly. The backslash `\` is the *escape character* — it tells Python *"the next character has a special meaning."*

Most common escape sequences:
- `\n` — newline (line break)
- `\t` — tab
- `\\` — a literal backslash
- `\"` — a literal double-quote (inside a double-quoted string)
- `\'` — a literal single-quote (inside a single-quoted string)

# Lesson 2: `.strip()`, `.upper()`, `.lower()` — Cleaning Methods

These are methods on strings that return a *new* string with some transformation applied. Important: strings are **immutable** — methods don't change the original, they hand back a new version.

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

```python
messy = "   Hello World   \n"

print(f"Original: {messy!r}")
print(f"Stripped: {messy.strip()!r}")
print(f"Upper:    {messy.upper()!r}")
print(f"Lower:    {messy.lower()!r}")

# The original is unchanged — methods return NEW strings
print(f"After all that, messy is still: {messy!r}")
```

# Lesson 3: `.split()` and `.join()` — Strings ↔ Lists

`.split()` turns a string into a list (breaking on a separator). `.join()` turns a list into a string (gluing with a separator). They're inverses of each other.

### 3a — `.split()`: string → list

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

```python
sentence = "Python is genuinely useful"

words = sentence.split()         # default: split on whitespace
print(words)

csv_line = "apple,banana,cherry"
fruits = csv_line.split(",")    # split on commas
print(fruits)
```

### 3b — `.join()`: list → string

**🔮 Predict** — what do I think the next code cell will do?

*(write your prediction here, in plain English, before running the code)*

```python
words = ["Python", "is", "genuinely", "useful"]

rejoined = " ".join(words)       # glue with spaces
print(rejoined)

newlined = "\n".join(words)      # glue with newlines
print(newlined)

compact = "".join(words)         # glue with nothing
print(compact)
```
