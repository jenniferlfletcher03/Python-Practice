```python
from llm_helpers import get_llm_response

with open("raw_notes/note-01.txt", "r") as f:
    contents = f.read()
```

```python
prompt = f"""You are helping clean up a research note from an autoethnographic study of human-AI interaction. The note is below in its original form. Your task is to reorganize the content into a consistent structure without changing the substance of what was written.

Output the cleaned note in exactly this format:

DATE: [extract the date from the note if findable, otherwise write: Unknown]

TAGS: [ comma-separated list of any tags found in the note, otherwise write: None]

SESSION CAPTURE: [Reformat the session capture content as a numbered list of distinct observations, one per line. The original may be in paragraphs, sections, or other formats - convert it to a numbered list regardless. If the note contains no session capture, write: Not included]

USER NOTE: [Preserve the user's own written note as-is. If not present, write: Not included]

CACHE'S INPUT: [Preserve Cache's contribution as-is. If not present, write: Not included]

ADDITIONAL OBSERVATIONS: [Anything in the note that doesn't fit the above sections. If none, write: None]

Guidelines:
- Preserve the original wording. Light cleanup of typos or broken fragments is fine, but do not paraphrase or summarize.
- If a section is missing from the original, explicitly write "Not included" or "None"- do not omit the section header.
- Cache is the name the user gave to their ChatGPT instance.

Here is the note to clean up:

{contents}

Thank you for your careful work on this - I appreciate your effort!
"""
```

```python
cleaned_note = get_llm_response(prompt)
print(cleaned_note)
```
